## Binomial Distribution

A binomial distribution describes the number of successes in a fixed number of independent trials.

### Conditions

1. The number of trials, **n**, is fixed.
2. Each trial has two possible outcomes: success or failure.
3. Trials are independent.
4. The probability of success, **p**, stays the same for every trial.

“Success” simply means the outcome we are counting—for example, heads.

### Probability Formula (PMF)

If X is the number of successes:

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$

where:

- **n** = total number of trials
- **k** = desired number of successes, from 0 to n
- **p** = probability of success on each trial
- **1 − p** = probability of failure on each trial
- **n − k** = number of failures

The combination is:

$$
\binom{n}{k}=\frac{n!}{k!(n-k)!}
$$

### Meaning of the Formula

$$
\text{Probability}
=
\text{Number of arrangements}
\times
\text{Probability of each arrangement}
$$

- The combination counts the arrangements containing exactly k successes.
- The remaining factors give the probability of one such arrangement.
- These arrangements are mutually exclusive and equally probable under the binomial conditions, so we add their probabilities by multiplying.

### Example: Exactly Two Heads in Four Tosses

Suppose independent coin tosses have:

- P(Heads) = 0.6
- P(Tails) = 0.4
- n = 4
- k = 2

$$
P(X=2)=\binom{4}{2}(0.6) * * 2 *(0.4)* *2
$$

$$
=6\times0.36\times0.16
=0.3456
=34.56\%
$$

There are six arrangements:

HHTT, HTHT, HTTH, THHT, THTH, TTHH

### Python

import math

probability = math.comb(4, 2) * 0.6^2 * 0.4^2

print(probability)

Result: approximately 0.3456.


### Summary

A binomial PMF answers:

“What is the probability of exactly k successes in n independent trials with a constant success probability p?”

Counting favourable outcomes and dividing by total outcomes works only when the elementary outcomes are equally likely. The binomial formula also handles biased coins.

## Continuous Probability: PDF, CDF and Uniform Distribution

### 1. Continuous Random Variable

A continuous random variable can take any value within its support.

Example: delivery time X, modeled as uniformly distributed between 20 and 40 minutes.

$$
X \sim \text{Uniform}(20,40)
$$

Equal-length intervals within this range have equal probabilities.

### 2. PDF: Probability Density Function

The PDF describes probability density. Its height at a point is not the probability of that exact value.

A valid PDF satisfies:

- Density is nonnegative.
- Total area under the density curve equals 1.

Probability over an interval is the area under the PDF:

$$
P(a \le X \le b)=\int_a^b f_X(x)\,dx
$$

For our uniform distribution:

$$
f_X(x)=\frac{1}{40-20}=0.05
$$

This density applies between 20 and 40; outside that range it is zero.

The density has units of “per minute.” Probability has no units.

### 3. Uniform Interval Probability

Because the density is constant:

$$
\text{Probability}=\text{Interval width}\times\text{Density}
$$

Example:

$$
P(24\le X\le30)
=(30-24)\times\frac1{20}
=0.30=30\%
$$

### 4. Exact Points and Endpoints

In this continuous model:

$$
P(X=x)=0
$$

A point has zero width and therefore zero area. Zero probability does not mean that value is impossible.

Consequently:

$$
P(X>30)=P(X\ge30)
$$

and:

$$
P(24<X<30)=P(24\le X\le30)
$$

A recorded time rounded to “30 minutes” represents an interval, which may have positive probability.

### 5. CDF: Cumulative Distribution Function

The CDF gives the probability up to and including a cutoff:

$$
F_X(x)=P(X\le x)
$$

For Uniform(a,b):

$$
F_X(x)=
\begin{cases}
0 & x<a\\
\frac{x-a}{b-a} & a\le x\le b\\
1 & x>b
\end{cases}
$$

The CDF never decreases and stays between 0 and 1.

### 6. Below, Above and Between

Below a cutoff:

$$
P(X\le b)=F_X(b)
$$

Above a cutoff:

$$
P(X>b)=1-F_X(b)
$$

Between two cutoffs:

$$
P(a\le X\le b)=F_X(b)-F_X(a)
$$

These endpoint equalities apply to continuous distributions.

### 7. Python Equivalents

```python
from scipy import stats

# loc = lower limit
# scale = upper limit minus lower limit
delivery = stats.uniform(loc=20, scale=20)

# Density at 30, not point probability
delivery.pdf(30)  # 0.05

# At or below 30 minutes
delivery.cdf(30)  # 0.50

# More than 35 minutes
delivery.sf(35)   # 0.25

# Between 24 and 30 minutes
delivery.cdf(30) - delivery.cdf(24)  # 0.30
```

### Summary

- PDF gives density; interval area gives probability.
- CDF gives cumulative probability up to a cutoff.
- SF gives probability strictly above a cutoff.
- Exact points have zero probability in a continuous distribution.
- Uniform distributions assign equal probabilities to equal-length intervals within their support.

## Expected Value

Expected value is the probability-weighted average of a random variable.

### Discrete Variable

Multiply each possible value by its probability, then add:

$$
E[X]=\sum_x xP(X=x)
$$

### Continuous Variable

Use the probability density to accumulate contributions:

$$
E[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx
$$

### Uniform Distribution

For X uniformly distributed between a and b:

$$
E[X]=\frac{a+b}{2}
$$

For delivery times between 20 and 40 minutes:

$$
E[X]=\frac{20+40}{2}=30\text{ minutes}
$$

### Interpretation

Under independent repeated observations from the same model,
the sample average tends toward the expected value as sample size grows.

This does not mean:

- Every delivery takes 30 minutes.
- Individual delivery times become less variable.
- The mean alone fully describes service performance.

Expected value need not be a possible individual outcome.
For example, a fair die has expected value 3.5.

### Python

    from scipy import stats

    delivery = stats.uniform(loc=20, scale=20)
    delivery.mean()  # 30.0

### Summary

Expected value describes the distribution's average.
Variance and standard deviation describe its spread.

## Variance and Standard Deviation

### Meaning

Variance measures the probability-weighted average squared distance
from the mean.

Standard deviation is the square root of variance.
It describes spread in the original units.

### Population Calculation by Hand

For the complete population: 28, 30, 32

Mean:

$$
\mu=\frac{28+30+32}{3}=30
$$

| Value | Deviation from 30 | Squared deviation |
|---:|---:|---:|
| 28 | -2 | 4 |
| 30 | 0 | 0 |
| 32 | +2 | 4 |

Population variance:

$$
\sigma^2=\frac{4+0+4}{3}=\frac83
$$

Population SD:

$$
\sigma=\sqrt{\frac83}\approx1.633
$$

Squaring prevents positive and negative deviations from cancelling.

For estimating population variance from a sample, the usual sample
formula divides the squared-deviation sum by n-1 instead of N.

### Continuous Uniform Distribution

For X uniformly distributed between a and b:

$$
\mu=\frac{a+b}{2}
$$

$$
\sigma^2=\frac{(b-a)^2}{12}
$$

$$
\sigma=\sqrt{\frac{(b-a)^2}{12}}
$$

The denominator 12 is a derived constant specific to continuous
uniform distributions. It is not the number of observations.

### Example: Uniform Delivery Times from 10 to 16 Minutes

Mean:

$$
\mu=\frac{10+16}{2}=13\text{ minutes}
$$

Variance:

$$
\sigma^2=\frac{(16-10)^2}{12}=3\text{ minutes}^2
$$

SD:

$$
\sigma=\sqrt3\approx1.732\text{ minutes}
$$

### Python

    from scipy import stats

    delivery = stats.uniform(loc=10, scale=6)

    delivery.mean()  # 13.0
    delivery.var()   # 3.0
    delivery.std()   # approximately 1.732

### Interpretation

- SD describes spread; it is not a maximum distance from the mean.
- Variance has squared units; SD has the original units.
- Equal means do not imply equal spreads.
- Multiplying all values by c multiplies SD by |c|
  and variance by c squared.

### Connection to Z-Scores

$$
z=\frac{x-\mu}{\sigma}
$$

Rearranging:

$$
x=\mu+z\sigma
$$

Z-scores do not require normality merely to calculate them.
The 68–95–99.7 rule applies to normal distributions.

## Bernoulli Versus Binomial

### Bernoulli Distribution

A Bernoulli variable records the result of ONE trial:

- 1 = the event occurs (“success”)
- 0 = the event does not occur (“failure”)

“Success” means the outcome being counted, even if it is undesirable.

Parameter: p, the probability of success.

$$
P(Y=1)=p,\qquad P(Y=0)=1-p
$$

PMF:

$$
P(Y=y)=p^y(1-p)^{1-y},\qquad y\in\{0,1\}
$$

### Binomial Distribution

A binomial variable counts successes across n trials.

Conditions:

1. Fixed number of trials, n.
2. Two outcomes per trial.
3. Independent trials.
4. Constant success probability, p.

Possible counts: 0, 1, ..., n.

$$
P(X=k)=\binom nk p^k(1-p)^{n-k}
$$

Bernoulli is the special case of binomial with n = 1.

### Example

Each order independently has a 10% probability of being incorrect.

- One order’s incorrect/correct indicator: Bernoulli(p=0.10).
- Number incorrect among eight orders: Binomial(n=8, p=0.10).

Probability of zero incorrect orders:

$$
P(X=0)=\binom80(0.1)^0(0.9)^8
=0.43046721\approx43.05\%
$$

### Python

```python
import math
from scipy import stats

# One order is incorrect
stats.bernoulli.pmf(1, p=0.10)

# Zero incorrect orders among eight
manual = math.comb(8, 0) * 0.1**0 * 0.9**8
scipy_result = stats.binom.pmf(0, n=8, p=0.10)
```

Tiny differences in trailing decimal digits can result from
floating-point rounding.

### Key Distinction

Both distributions calculate probabilities.

Bernoulli describes one binary trial.
Binomial counts successes across a fixed number of independent
binary trials with the same success probability.

````python
print(stats.bernoulli.pmf(1, p=0.20))  
print(stats.bernoulli.pmf(0, p=0.20))
````

## Bernoulli Distribution

### Definition

A Bernoulli distribution describes ONE trial with two possible outcomes.

Define the random variable:

- Y = 1 when the event being counted occurs.
- Y = 0 when it does not occur.

The event called “success” need not be desirable.

### Parameter and Possible Values

- p = probability of success
- 1 − p = probability of failure
- Possible values of Y: 0 and 1
- 0 ≤ p ≤ 1

### Probability Mass Function

$$
P(Y=y)=p^y(1-p)^{1-y},\qquad y\in\{0,1\}
$$

For success:

$$
P(Y=1)=p
$$

For failure:

$$
P(Y=0)=1-p
$$

### Example: One Customer’s Advertisement Click

A customer has a 20% probability of clicking an advertisement.

| Y | Outcome | Probability |
|---:|---|---:|
| 0 | No click | 0.80 |
| 1 | Click | 0.20 |

The probabilities sum to 1.

### Python

```python
from scipy import stats

stats.bernoulli.pmf(1, p=0.20)  # 0.20
stats.bernoulli.pmf(0, p=0.20)  # approximately 0.80
```

### Connection to Binomial

Bernoulli describes one binary trial.

Binomial counts successes across a fixed number of independent
Bernoulli trials with the same success probability.

Bernoulli(p) is equivalent to Binomial(n=1, p=p).

Independence between trials is a condition for this binomial
construction, not a requirement for describing one Bernoulli variable.

````python
print(stats.poisson.pmf(0,mu=2))
print(stats.poisson.pmf(0,2))
````

## Poisson Distribution

### Definition

A Poisson distribution models the number of events occurring
within a fixed interval of time, space or another exposure.

Example: number of orders arriving during five minutes.

Possible counts: 0, 1, 2, 3, ...
There is no fixed upper limit in the model.

### Expected Count: Lambda

$$
\lambda=\text{average rate}\times\text{interval length}
$$

Keep units consistent.

Example: 4 orders per minute over 3 minutes:

$$
\lambda=4\times3=12
$$

This is an expected count, not a guaranteed count.

### Assumptions for a Constant-Rate Poisson Process

- The event rate is constant over the interval.
- Counts in nonoverlapping intervals are independent.
- Events occur individually rather than in simultaneous batches.

An average rate alone does not establish that a Poisson model fits.

### Probability Mass Function

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
$$

- k = exact count of interest
- lambda = expected count for the chosen interval
- e ≈ 2.71828
- 0! = 1

### Example: Two Orders per Minute

For a one-minute interval, lambda = 2.

Exactly zero orders:

$$
P(X=0)=\frac{e^{-2}2^0}{0!}
=e^{-2}\approx0.1353
$$

At most two orders:

$$
P(X\le2)=P(X=0)+P(X=1)+P(X=2)
\approx0.6767
$$

More than two orders:

$$
P(X>2)=1-P(X\le2)\approx0.3233
$$

### Python

```python
import math
from scipy import stats

# SciPy calls the expected-count parameter "mu".
# Here, mu represents lambda.

manual_zero = math.exp(-2) * 2**0 / math.factorial(0)

stats.poisson.pmf(0, mu=2)  # Exactly 0: approximately 0.1353
stats.poisson.cdf(2, mu=2)  # At most 2: approximately 0.6767
stats.poisson.sf(2, mu=2)   # More than 2: approximately 0.3233

# At least 2 includes 2, so exclude only 0 and 1:
stats.poisson.sf(1, mu=2)
```

### Binomial Versus Poisson

| Binomial | Poisson |
|---|---|
| Fixed number of trials | Fixed observation interval |
| Counts successes among n trials | Counts events during the interval |
| Parameters: n and p | Parameter: lambda |
| Maximum count is n | No fixed upper limit |

### Summary

Choose the observation interval first.
Convert the rate into its expected count.
Use PMF for exactly, CDF for at most, and SF for more than.

````python
print(stats.binom.mean(10,0.2), stats.binom.var(10,0.2), stats.binom.std(10,0.2))
````

## Mean, Variance and SD: Bernoulli, Binomial and Poisson

Standard deviation is always the square root of variance.

| Distribution | Mean | Variance | SD |
|---|---|---|---|
| Bernoulli(p) | p | p(1-p) | sqrt(p(1-p)) |
| Binomial(n,p) | np | np(1-p) | sqrt(np(1-p)) |
| Poisson(lambda) | lambda | lambda | sqrt(lambda) |

### Bernoulli

One binary trial, coded 1 for success and 0 for failure:

$$
E[Y]=0(1-p)+1p=p
$$

Variance is the probability-weighted squared distance from the mean:

$$
\operatorname{Var}(Y)
=(0-p)^2(1-p)+(1-p)^2p
=p(1-p)
$$

$$
\operatorname{SD}(Y)=\sqrt{p(1-p)}
$$

Example: p = 0.20

- Mean = 0.20
- Variance = 0.16
- SD = 0.40

The expected value need not be a possible individual outcome.

### Binomial

A binomial count sums n independent Bernoulli trials with the same p.

Means add. Variances also add because the trials are independent.
Do not add their SDs.

$$
E[X]=np
$$

$$
\operatorname{Var}(X)=np(1-p)
$$

$$
\operatorname{SD}(X)=\sqrt{np(1-p)}
$$

Example: n = 10, p = 0.20

- Mean = 2
- Variance = 1.6
- SD ≈ 1.2649

### Poisson

For the selected observation interval:

$$
E[X]=\lambda
$$

$$
\operatorname{Var}(X)=\lambda
$$

$$
\operatorname{SD}(X)=\sqrt{\lambda}
$$

Example: lambda = 2

- Mean = 2
- Variance = 2
- SD ≈ 1.4142

Equal numerical mean and variance is a Poisson property,
not a property of every count distribution.

### Python

```python
from scipy import stats

models = {
    "Bernoulli": stats.bernoulli(p=0.20),
    "Binomial": stats.binom(n=10, p=0.20),
    "Poisson": stats.poisson(mu=2),
}

for name, model in models.items():
    print(name, model.mean(), model.var(), model.std())
```

### Interpretation

- Mean describes the expected value.
- Variance describes average squared distance from that mean.
- SD describes spread in the original units.
- Neither the mean nor SD guarantees an individual result.

## Geometric Distribution

### Definition

A geometric variable counts trials until the first success.

Here, X INCLUDES the successful trial.

Possible values: 1, 2, 3, ...

### Conditions

- Each trial has two outcomes: success or failure.
- Trials are independent.
- Success probability p stays constant.
- We continue until the first success.

### PMF: First Success on Trial k

The first k−1 trials fail, then trial k succeeds:

$$
P(X=k)=(1-p)^{k-1}p
$$

No combination factor is needed: the required order is specified.

Example: p = 0.20, first success on trial 4:

$$
P(X=4)=0.8^3(0.2)=0.1024
$$

### CDF: First Success Within k Trials

The complement is failure on all first k trials:

$$
P(X\le k)=1-(1-p)^k
$$

Example:

$$
P(X\le4)=1-0.8^4=0.5904
$$

### Mean, Variance and SD

$$
E[X]=\frac1p
$$

$$
\operatorname{Var}(X)=\frac{1-p}{p^2}
$$

$$
\operatorname{SD}(X)=\sqrt{\frac{1-p}{p^2}}
$$

For p = 0.25:

- Mean = 4 trials
- Variance = 12 trials²
- SD ≈ 3.464 trials

An expected wait of four trials does not guarantee success
on the fourth trial.

### Python

```python
from scipy import stats

model = stats.geom(p=0.20)

model.pmf(4)  # First success on trial 4: 0.1024
model.cdf(4)  # First success within 4 trials: 0.5904

model.mean()  # 5
model.var()   # 20
model.std()   # approximately 4.472
```

### Counting Convention

Some resources count FAILURES before the first success instead.

If W = X−1, W can be 0, 1, 2, ...

Its mean is (1−p)/p.
Its variance is the same as X because subtracting 1 changes
location, not spread.

SciPy's geom uses X: trials INCLUDING the first success.

### Binomial Versus Geometric

- Binomial: fixed number of trials; count successes.
- Geometric: continue until first success; count trials.

## Hypergeometric Distribution

A hypergeometric distribution describes the number of successes in a fixed-size random sample drawn **without replacement** from a finite population.

“Success” means the type of item we are counting, such as a defective part.

### Conditions

1. The population size is fixed.
2. Each item belongs to one of two categories: success or failure.
3. The number selected is fixed.
4. Selection is random and without replacement.

Because items are not replaced, successive draws are dependent.

### Mathematical Notation

- N = total population size
- K = number of successes in the population
- n = number of items selected
- k = number of successes selected
- X = random variable counting successes selected

The possible values of X satisfy:

$$
\max(0,n-(N-K)) \leq k \leq \min(n,K)
$$

You cannot select more successes than exist, or more items than your sample contains.

### Probability Formula — PMF

$$
P(X=k)=
\frac{\binom{K}{k}\binom{N-K}{n-k}}
{\binom{N}{n}}
$$

Meaning:

$$
\frac{
\text{Ways to choose the successes}
\times
\text{Ways to choose the failures}
}{
\text{All possible samples}
}
$$

The samples counted in the denominator are equally likely under random selection.

### Example

A box contains 10 parts: 3 defective and 7 good.
Select 4 parts randomly without replacement.
Let X count defective parts selected.

Exactly one defective:

$$
P(X=1)=
\frac{\binom31\binom73}{\binom{10}4}
=\frac{3\times35}{210}
=0.5
$$

At most one defective:

$$
P(X\leq1)=P(X=0)+P(X=1)
=\frac{35}{210}+0.5
=\frac23
$$

At least two defective:

$$
P(X\geq2)=1-P(X\leq1)=\frac13
$$

### Mean, Variance and Standard Deviation

Let:

$$
p=\frac KN
$$

Then:

$$
E[X]=np=n\frac KN
$$

$$
\operatorname{Var}(X)=
np(1-p)\frac{N-n}{N-1}
$$

$$
SD(X)=\sqrt{\operatorname{Var}(X)}
$$

The variance formula above assumes N > 1.

For our example:

$$
E[X]=4(0.3)=1.2
$$

$$
\operatorname{Var}(X)=
4(0.3)(0.7)\frac69=0.56
$$

$$
SD(X)=\sqrt{0.56}\approx0.7483
$$

A mean of 1.2 means the average defective count across many
repeated samples approaches 1.2. Each individual count remains
a whole number. Restore the population before each repeated sample.

### Why the Extra Variance Factor?

$$
\frac{N-n}{N-1}
$$

This corrects for sampling without replacement.

Compared with independent binomial trials with the same n and p,
sampling without replacement reduces the variability of the count.

If we select the entire population, n = N:

$$
\operatorname{Var}(X)=0
$$

The total success count is then known with certainty.

### Python — SciPy Parameter Mapping

SciPy uses different parameter names:

| Mathematical notation | Meaning | SciPy parameter |
|---|---|---|
| N | Population size | M |
| K | Successes in population | n |
| n | Sample size | N |

Use keyword arguments to avoid mixing them up.

````python
from scipy import stats

parts = stats.hypergeom(M=10, n=3, N=4)

# Exactly 1 defective
print(parts.pmf(1))       # 0.5

# At most 1 defective
print(parts.cdf(1))       # approximately 0.6667

# At least 2 defective: strictly more than 1
print(parts.sf(1))        # approximately 0.3333

# Moments
print(parts.mean())      # 1.2
print(parts.var())       # 0.56
print(parts.std())       # approximately 0.7483
````

### Integer-Count Probability Shortcuts

For integer boundaries a and b:

- Exactly k: pmf(k)
- At most k: cdf(k)
- More than k: sf(k)
- At least k: sf(k - 1)
- Between a and b, inclusive: cdf(b) - cdf(a - 1)

### Binomial Versus Hypergeometric

| Binomial | Hypergeometric |
|---|---|
| Independent trials | Dependent draws without replacement |
| Constant success probability | Conditional success probability changes |
| Fixed number of trials | Fixed sample from a finite population |

### Summary

Use hypergeometric when counting successes in a random,
fixed-size sample drawn without replacement from a finite population
with a known number of successes.

## Exponential Distribution

The exponential distribution models waiting time until the next
event in a Poisson process: events occur independently at a
constant average rate.

### Poisson Connection

- Poisson: number of events within a fixed interval.
- Exponential: waiting time until the next event.

An average arrival rate alone does not establish that either
model fits; check the process assumptions.

### Rate and Mean

Let λ be the event rate and T the waiting time.

$$
E[T]=\frac{1}{\lambda}
$$

Rate and time must use matching units.

Example: 3 arrivals per minute means an average wait of
1/3 minute = 20 seconds.

### Probabilities

For t ≥ 0:

More than t — survival function:

$$
P(T>t)=e^{-\lambda t}
$$

At most t — cumulative distribution function:

$$
F(t)=P(T\leq t)=1-e^{-\lambda t}
$$

Between a and b, where 0 ≤ a < b:

$$
P(a<T\leq b)=F(b)-F(a)
=e^{-\lambda a}-e^{-\lambda b}
$$

Because T is continuous:

$$
P(T=t)=0
$$

Including or excluding an exact endpoint does not change an
interval probability.

### Probability Density — PDF

$$
f(t)=\lambda e^{-\lambda t},\qquad t\geq0
$$

The PDF gives curve height, not probability at an exact value.
Probability is area under the density curve over an interval.

The distribution is right-skewed: many short waits and some
long waits.

### Mean, Variance and SD

$$
E[T]=\frac1\lambda
$$

$$
\operatorname{Var}(T)=\frac1{\lambda^2}
$$

$$
SD(T)=\frac1\lambda
$$

For an exponential distribution, mean and SD are equal.
Variance uses squared time units.

### Python Mapping

SciPy uses stats.expon.

For waiting from time zero:

- loc = 0, the default
- scale = 1/λ = mean waiting time

Do not pass the rate as scale.

## Beta Distribution

A beta distribution models a continuous proportion or probability
between 0 and 1.

Examples:
- Uncertainty about a factory's paint-defect probability.
- Variation in underlying purchase rates across car dealerships.

These are different interpretations: uncertainty about one rate
versus variation across multiple rates.

### Parameters

α > 0 and β > 0 are shape parameters.
They are NOT the probabilities p and 1 − p.

### Mean, Variance and SD

$$
\mu=\frac{\alpha}{\alpha+\beta}
$$

$$
\operatorname{Var}(X)=
\frac{\alpha\beta}
{(\alpha+\beta)^2(\alpha+\beta+1)}
$$

$$
SD(X)=\sqrt{\operatorname{Var}(X)}
$$

Holding the mean fixed, increasing α + β reduces the spread.

### Shape

- α = β = 1: uniform between 0 and 1.
- α = β > 1: symmetric around 0.5.
- α > β: mean above 0.5.
- α < β: mean below 0.5.
- Both below 1: U-shaped.

### Updating an Unknown Probability

If p starts with a Beta(α, β) distribution and observations
are independent Bernoulli trials with the same p:

$$
\alpha_{\text{new}}=\alpha_{\text{old}}+\text{successes}
$$

$$
\beta_{\text{new}}=\beta_{\text{old}}+\text{failures}
$$

“Success” means the outcome being counted—even a defect.

Example:
Starting Beta(2,8), then observing 1 defective and 9 nondefective:

$$
p\mid\text{data}\sim\operatorname{Beta}(3,17)
$$

Updated mean probability = 3/20 = 15%.

Starting parameters must be justified; they are not automatically
(1,1), and need not be whole numbers.

### Probability Versus Density

- CDF: probability of being at or below a threshold.
- SF: probability of being above a threshold.
- PDF: density height, not probability at an exact value.
- For this continuous distribution, P(X = x) = 0.

### Beta Versus Binomial

- Beta models a probability p.
- Binomial models a success count given a fixed p.

## Gamma Distribution

Gamma models positive continuous quantities, including waiting
times and amounts.

### Connection to Poisson and Exponential

For a Poisson process with constant rate λ:

- Poisson models the count within an interval.
- Exponential models time until the first arrival.
- Gamma with integer shape k models time until the kth arrival.

This integer-shape case is also called Erlang.

It is the sum of k independent exponential waiting gaps
with the same rate.

### Shape and Scale

- k > 0: shape
- θ > 0: scale
- In the arrival example, θ = 1/λ.

Gamma also permits noninteger shapes, but those do not represent
a fractional number of arrivals.

### Mean, Variance and SD

$$
E[T]=k\theta
$$

$$
\operatorname{Var}(T)=k\theta^2
$$

$$
SD(T)=\sqrt{k}\theta
$$

Independent gaps have additive means and variances.
Do not square the total mean to obtain gamma variance.

### Example

Cars arrive at 4 per minute.
Time until the third arrival:

- Shape k = 3.
- Scale θ = 15 seconds.
- Mean = 45 seconds.
- Variance = 675 seconds².
- SD ≈ 25.98 seconds.

### Probability Connection

For positive integer k:

$$
P(T_k\leq t)=P(N(t)\geq k)
$$

The kth arrival has happened by time t exactly when
at least k arrivals have occurred by then.

For Poisson, the expected count is λt.
For gamma, the time scale is 1/λ.
Keep rate and time units consistent.

### Shape and Support

- Possible waiting times are nonnegative.
- Shape 1 gives an exponential distribution.
- Shape greater than 1 gives a peak away from zero.
- Increasing shape makes the distribution less right-skewed.

### Python

SciPy uses:
- a = shape
- scale = θ
- loc = 0 for waiting from now

CDF gives probability at or below a time.
SF gives probability above a time.
PDF gives density height, not exact-point probability.
